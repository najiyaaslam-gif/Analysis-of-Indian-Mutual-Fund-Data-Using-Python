**Handling Outliers**

```python
# Detect outliers using IQR

exclude_cols = [
    'id',
    'no_of_segregated_portfolios',
    'net_aum_in_segregated_portfolios'
]

numericcols = [
    col for col in df.select_dtypes(include='number').columns
    if col not in exclude_cols
]

for col in numericcols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print("\nColumn:", col)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

**Histogram of Net AUM**

```python
plt.figure(figsize=(10, 5))

plt.hist(df['net_aum'], bins=20, color="purple",edgecolor='black')

plt.title('Distribution of Net AUM')
plt.xlabel('Net AUM')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

The histogram shows a right-skewed distribution of Net AUM. Most observations fall within the lower value ranges, while a smaller number of observations have substantially higher Net AUM values, forming a long right tail.


**Net AUM Box plot**

```python
plt.figure(figsize=(12, 5))

ax = sns.boxplot(x=df['net_aum'])

Q1 = df['net_aum'].quantile(0.25)
median = df['net_aum'].median()
Q3 = df['net_aum'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

ax.axvline(Q1, color='green', linestyle='--',
           label=f'Q1 = {Q1:.2f}')
ax.axvline(median, color='red', linestyle='--',
           label=f'Median = {median:.2f}')
ax.axvline(Q3, color='purple', linestyle='--',
           label=f'Q3 = {Q3:.2f}')

plt.title('Boxplot of Net AUM')
plt.xlabel('Net AUM')
plt.legend()
plt.show()
```

The boxplot shows the median and spread of Net AUM values. The interquartile range is wide, and several observations lie beyond the upper whisker, indicating potential high-value outliers.
